# Changing Dictionary Items

## 1. Change a value using its key

Use `dictionary[key] = new_value` to change the value of an existing item. In this example, we change `"age"` from `20` to `21`.

If the key does not exist, this assignment adds a new item instead.

```python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 20,
    "city": "Prishtina",
}

# Change an existing value.
person["age"] = 21

print(person)
```

## 2. Change or add items with update()

The `update()` method can change multiple values and add new items in one call. It accepts a dictionary (or another mapping), an iterable of `(key, value)` pairs, or keyword arguments.

In this example:

- The existing `"surname"` value changes from `"Ceku"` to `"Çeku"`.
- The new `"pets"` key is added with the value `"Riki&Roki"`.

`update()` changes the original dictionary and returns `None`.

```python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 20,
    "city": "Prishtina",
}

# Update an existing item and add a new one.
person.update({
    "surname": "Çeku",
    "pets": "Riki&Roki",
})

print(person)
```

## Quick recap

| Syntax | What it does |
| --- | --- |
| `person["age"] = 21` | Changes one value, or adds the key if it is missing. |
| `person.update({"surname": "Çeku", "pets": "Riki&Roki"})` | Updates existing keys and adds new keys in one call. |

**Remember:** Both approaches change the original dictionary. Each example above starts with a fresh `person` dictionary, so the age in the second example is still `20`.
