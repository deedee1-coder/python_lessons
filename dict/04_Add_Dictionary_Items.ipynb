# Adding Dictionary Items

## 1. Add an item using a new key

Use `dictionary[new_key] = value` to add a key-value pair. In this example, we add the key `"surname"` with the value `"Ceku"`.

If the key already exists, its value is updated instead.

```python
person = {
    "name": "Dea",
    "age": 21,
}

print("Before:", person)

# Add a new key-value pair.
person["surname"] = "Ceku"

print("After:", person)
```

## 2. Add items with update()

The `update()` method adds items whose keys are missing and updates values whose keys already exist. You can pass a dictionary containing one or more key-value pairs.

Here, `person.update({"surname": "Ceku"})` adds the same item as the first example. Each example starts with a fresh dictionary.

```python
person = {
    "name": "Dea",
    "age": 21,
}

print("Before:", person)

# Add a new item with update().
person.update({"surname": "Ceku"})

print("After:", person)
```

## Quick recap

| Syntax | What it does |
| --- | --- |
| `person["surname"] = "Ceku"` | Adds one item, or updates its value if the key already exists. |
| `person.update({"surname": "Ceku"})` | Adds or updates the items supplied in a dictionary. |

**Remember:** Both approaches change the original dictionary. New keys appear at the end in insertion order. The `update()` method returns `None`, so call it directly rather than assigning its result to `person`.
