# Accessing Items in a Python Dictionary

## 1. Access a value with square brackets

Use `dictionary[key]` to get the value associated with a key. For example, `person["name"]` returns `"Dea"`. A missing key raises a `KeyError`.

```python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 21,
}

person_name = person["name"]
print(person_name)
```

## 2. Access a value with get()

The `get()` **method** also returns the value for a key. If the key is missing, it returns `None` instead of raising a `KeyError`. You can supply a default value, such as `person.get("city", "Unknown")`.

```python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 21,
}

person_name = person.get("name")
print(person_name)
```

## 3. Get all keys with keys()

The `keys()` method returns a **view of all keys**, rather than a list. Use `list(person.keys())` if you need a list. The view reflects changes to the dictionary.

```python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 21,
}

person_keys = person.keys()
print(person_keys)
```

## 4. Add a new item

Assign a value to a new key using `dictionary[key] = value`. Here, we add `"city": "Prishtina"`. If the key already exists, this updates its value.

```python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 21,
}

person["city"] = "Prishtina"
print(person)
```

## 5. Get all values with values()

The `values()` method returns a **view of all values**. Use `list(person.values())` to convert it to a list. Like the keys view, it reflects dictionary changes.

```python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 21,
}

print(person.values())
```

## 6. Get key-value pairs with items()

The `items()` method returns a **view of key-value pairs**. Each pair is a tuple, such as `("name", "Dea")`. Use `list(person.items())` if you need a list of tuples.

```python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 21,
}

print(person.items())
```

## 7. Check whether a key exists

Use `key in dictionary` to check for a **key**. It returns `True` or `False`. This example checks for `"city"` in a fresh dictionary that does not contain that key.

```python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 21,
}

if "city" in person:
    print("Yes, the 'city' key exists in the dictionary.")
else:
    print("No, the 'city' key does not exist in the dictionary.")
```

## Quick recap

| Syntax | What it does |
| --- | --- |
| `person["name"]` | Gets a value; raises `KeyError` if the key is missing. |
| `person.get("name")` | Gets a value; returns `None` if the key is missing. |
| `person.keys()` | Returns a view of the keys. |
| `person["city"] = "Prishtina"` | Adds an item or updates an existing value. |
| `person.values()` | Returns a view of the values. |
| `person.items()` | Returns a view of key-value tuples. |
| `"city" in person` | Checks whether a key exists. |

**Remember:** String keys are case-sensitive: `"name"` and `"Name"` are different keys. Each code example above creates its own `person` dictionary.
