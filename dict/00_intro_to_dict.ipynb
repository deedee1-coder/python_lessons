# Introduction to Python Dictionaries

## 1. Creating a dictionary

A dictionary stores **key-value pairs** inside curly braces `{}`. Separate each key and value with a colon, and each pair with a comma.

In the example below, `"father"` is a **key** and `"Vegim"` is its **value**.

````python
family = {
    "father": "Vegim",
    "mother": "Liridona",
    "brother": "Endrit",
    "sister1": "Enea",
    "sister2": "Nina",
}

print(family)
````

## 2. Ordered

Dictionaries are **ordered**, **changeable**, and have **unique keys**.

Ordered means a dictionary remembers the order in which you added its keys. When you print it or loop through it, the items appear in that order.

````python
family_pets = {}

family_pets["dog1"] = "miki"
family_pets["dog2"] = "riki"
family_pets["dog3"] = "roki"

for key, value in family_pets.items():
    print(key, value)
````

**Output:**

```text
dog1 miki
dog2 riki
dog3 roki
```

This is **insertion order**: Python does not automatically sort the keys alphabetically. Regular dictionaries guarantee insertion order in Python 3.7 and later.

## 3. Changeable (mutable)

You can **update a value**, **add an item**, or **remove an item** after creating a dictionary.

````python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 20,
}

# Update a value.
person["age"] = 21

# Add an item.
person["city"] = "Prishtina"

print(person)

# Remove an item using its key.
del person["surname"]

print(person)
````

**Output from the changeable example:**

```text
{'name': 'Dea', 'surname': 'Ceku', 'age': 21, 'city': 'Prishtina'}
{'name': 'Dea', 'age': 21, 'city': 'Prishtina'}
```

## 4. No duplicate keys

Each key appears only once in a dictionary. If you repeat a key, the **last value replaces the earlier one**.

````python
person = {
    "name": "Dea",
    "surname": "Ceku",
    "age": 20,
    "age": "twenty",
}

print(person)
````

**Output:**

```text
{'name': 'Dea', 'surname': 'Ceku', 'age': 'twenty'}
```

Only one `"age"` key remains, with the value `"twenty"`. Python does not raise an error; it overwrites the earlier value.

Different keys **can have the same value**. For example, `{"Dea": 20, "Enea": 20}` is valid.

## Quick recap

| Property | Meaning |
| --- | --- |
| Key-value pairs | Each key identifies a value. |
| Ordered | Keys keep the order in which they were added. |
| Changeable | You can update, add, and remove items. |
| Unique keys | A repeated key replaces its earlier value. |
| Duplicate values allowed | Different keys can share the same value. |
